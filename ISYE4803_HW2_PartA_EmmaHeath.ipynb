```python
import numpy as np
import pandas as pd
import statsmodels.formula.api as smf
```

```python
data = pd.read_csv("roe01.txt", delim_whitespace=True, header=0)
data
```

```python
#Column Means
print(data.mean(numeric_only=True))
```

```python
#Correlation
print(data['ROEt'].corr(data['ROE']))
```

```python
#Simple Linear Model
fit = smf.ols("ROE ~ ROEt", data=data).fit()
print(fit.summary())
```
